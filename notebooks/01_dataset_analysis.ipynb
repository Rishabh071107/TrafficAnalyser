# 01. Traffic Dataset Exploratory Data Analysis (EDA)

This notebook analyzes the traffic dataset structure, vehicle class distributions, bounding box aspects ratios, and image quality metrics.

```python
import os
import yaml
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset configuration
with open('../configs/data.yaml', 'r') as f:
    data_config = yaml.safe_load(f)

print("Dataset Config:", data_config)
```

```python
# Count labels per class
labels_dir = os.path.join('..', data_config['path'], 'labels', 'train')
class_names = data_config['names']
class_counts = {name: 0 for name in class_names.values()}

if os.path.exists(labels_dir):
    for lbl_file in os.listdir(labels_dir):
        if lbl_file.endswith('.txt'):
            with open(os.path.join(labels_dir, lbl_file), 'r') as f:
                for line in f:
                    parts = line.strip().split()
                    if parts:
                        cls_id = int(parts[0])
                        cls_name = class_names.get(cls_id, f'class_{cls_id}')
                        class_counts[cls_name] = class_counts.get(cls_name, 0) + 1

df_counts = pd.DataFrame(list(class_counts.items()), columns=['Class', 'Count'])
print(df_counts)

plt.figure(figsize=(8, 4))
plt.bar(df_counts['Class'], df_counts['Count'], color='skyblue')
plt.title('Vehicle Class Distribution in Training Set')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()
```
