# 03. Model Testing & Video Inference Interactive Visualizer

Test trained vehicle detector, tracker, speed estimator, and counter on traffic video.

```python
import sys
import os
import cv2
from IPython.display import display, Image
sys.path.append('..')

from src.detection.detector import TrafficDetector
from src.tracking.tracker import TrafficTracker
from src.traffic.counter import VehicleCounter
from src.utils.visualization import Visualizer

weights_path = '../models/trained/best.pt'
video_path = '../inputs/videos/traffic.mp4'

detector = TrafficDetector(weights_path=weights_path if os.path.exists(weights_path) else 'yolo11n.pt')
tracker = TrafficTracker()
counter = VehicleCounter()
visualizer = Visualizer()

cap = cv2.VideoCapture(video_path)
ret, frame = cap.read()
cap.release()

if ret:
    dets = detector.detect(frame)
    tracks = tracker.update(dets)
    annotated = visualizer.draw_tracks(frame, tracks)
    _, encoded = cv2.imencode('.jpg', annotated)
    display(Image(data=encoded.tobytes()))
else:
    print('Video frame could not be opened.')
```
